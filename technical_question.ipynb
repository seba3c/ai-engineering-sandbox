# Objective

Ask a technical question to a model using OpenAI API and OpenAI SDK

```python
from IPython.display import Markdown, display, update_display

from llm_utils import get_llm_client, load_api_key

api_key = load_api_key()

system_prompt = "You are an expert Software Engineer."

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.1:8b'
```

Reply in stream mode

```python
llm_client = get_llm_client(api_key=api_key)

stream = llm_client.chat.completions.create(
    model=MODEL_GPT,
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": question},
    ],
    stream=True,
)
response = ""
display_handle = display(Markdown(""), display_id=True)
for chunk in stream:
    response += chunk.choices[0].delta.content or ''
    update_display(Markdown(response), display_id=display_handle.display_id)
```

Reply without streaming

```python
OLLAMA_BASE_URL = "http://localhost:11434/v1"

llm_client = get_llm_client(base_url=OLLAMA_BASE_URL)

response = llm_client.chat.completions.create(
    model=MODEL_LLAMA,
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": question},
    ],
)


display(Markdown(response.choices[0].message.content))
```
