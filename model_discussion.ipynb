```python
class Chatbot:

    def __init__(self, name, model,  system_prompt, user_prompt_placeholder) -> None:
        self.name = name
        self.model = model
        self.system_prompt = system_prompt
        self.user_prompt_placeholder = user_prompt_placeholder

    def get_user_prompt(self, *args, **kwargs):
        return self.user_prompt_placeholder.format(*args, **kwargs)

    
    def __str__(self) -> str:
        return f"{self.name} - {self.model}"

```

Alex chatbot

```python
system_prompt = """
You are Alex, a chatbot who is very argumentative; you disagree with anything in the conversation and you challenge everything, in a snarky way.
You are in a conversation with Blake and Charlie.
Reply in Spanish.
"""

user_prompt_placeholder = """
You are Alex, in conversation with Blake and Charlie.
The conversation so far is as follows: '{0}'
Now with this, respond with what you would like to say next, as Alex.
"""

alex = Chatbot("Alex", "claude-haiku-4-5", system_prompt, user_prompt_placeholder)
```

Charlie chatbot

```python
system_prompt = """
You are Charlie, a chatbot who analize questions and responses before answer with your own point of view.
You are in a conversation with Blake and Alex.
Reply in Spanish.
"""

user_prompt_placeholder = """
You are Charlie, in conversation with Blake and Alex.
The conversation so far is as follows: '{0}'
Now with this, respond with what you would like to say next, as Charlie.
"""

charlie = Chatbot("Charlie", "gpt-4.1-nano", system_prompt, user_prompt_placeholder)
```

Blake

```python
system_prompt = """
You are Blake, a chatbot who accepts other point of views without hesitation.
You are in a conversation with Charlie and Alex.
Reply in Spanish.
"""

user_prompt_placeholder = """
You are Blake, in conversation with Charlie and Alex.
The conversation so far is as follows: 
{0}
Now with this, respond with what you would like to say next, as Blake.
"""

blake = Chatbot("Blake", "ollama/llama3.1:8b", system_prompt, user_prompt_placeholder)
```

Discussion begins

```python
from IPython.display import Markdown
from litellm import completion

def call_chatbot(chatbot: Chatbot, messages: list):
    # print("\n".join(messages))
    user_prompt = chatbot.get_user_prompt("\n".join(messages))
    messages = [{"role": "system", "content": chatbot.system_prompt}, {"role": "user", "content": user_prompt}]
    response = completion(model=chatbot.model, messages=messages)
    reply = response.choices[0].message.content
    return reply


chatbots = [charlie, alex, blake]

messages = [
    # "I have a metal cup with no bottom and a sealed top. How can I use this cup?"
    "if a dog choose shoes what shoes does he choose?"
]

display(Markdown(f"### Question:\n{messages[0]}\n"))

for i in range(3):
    for chatbot in chatbots:
        reply = call_chatbot(chatbot, messages)
        messages.append(f"[{chatbot.name}]: {reply}")
        display(Markdown(f"### {chatbot.name} - {chatbot.model}\n{reply}"))
```
