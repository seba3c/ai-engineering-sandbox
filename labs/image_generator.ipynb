### Objective

Generate a custom image using a llm that support image generation

```python
from llm_utils import load_api_keys, get_llm_client

load_api_keys()
```

Generates an image

```python
import base64
from io import BytesIO
from PIL import Image

llm_client = get_llm_client()

def artist(city):
    image_response = llm_client.images.generate(
            model="gpt-image-1-mini",
            prompt=f"An image representing a vacation in {city}, showing tourist spots and everything unique about {city}, in a vibrant pop-art style",
            size="1024x1024",
            n=1,
        )
    image_base64 = image_response.data[0].b64_json
    image_data = base64.b64decode(image_base64)
    return Image.open(BytesIO(image_data))
```

Display the image generator using gradio

```python
import gradio as gr

# Fixed list of selectable cities
CITIES = [
    "Paris",
    "Tokyo",
    "New York",
    "Rome",
    "Barcelona",
    "Bangkok",
    "Sydney",
    "Cairo",
    "Rio de Janeiro",
    "Cape Town",
]

with gr.Blocks(title="Vacation Artist") as demo:
    gr.Markdown(
        "# 🎨 Vacation Artist\n"
        "Pick a city and generate a vibrant pop-art travel poster for it."
    )
 
    with gr.Row():
        city_dropdown = gr.Dropdown(
            choices=CITIES,
            value=CITIES[0],
            label="City",
        )
        generate_btn = gr.Button("Generate Image", variant="primary")
 
    output_image = gr.Image(label="Generated Image", type="pil")
 
    generate_btn.click(
        fn=artist,
        inputs=city_dropdown,
        outputs=output_image,
    )

demo.launch()
```
